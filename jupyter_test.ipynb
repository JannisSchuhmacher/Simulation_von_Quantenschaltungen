# Pauli-Matrizen testen

Dieses Notebook verwendet dieselben Funktionen wie `playground/test.py` und überprüft die Relation $\sigma_x\sigma_y = i\sigma_z$.

```python
import numpy as np

from simulation_von_quantenschaltungen import hello
from simulation_von_quantenschaltungen.pauli_matrices import (
	pauli_x,
	pauli_y,
	pauli_z,
)

print(hello())

left_side = pauli_x() @ pauli_y()
right_side = 1j * pauli_z()

assert np.allclose(left_side, right_side)
print("sigma_x sigma_y = i sigma_z: verified")
```
